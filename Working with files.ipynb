## Getting data from a file

### Opening a file


Getting data out of a file is a two step process: open then read.

````python
my_file = open("weather.txt")
````

`open()` is a function that takes one string argument - the name of the file - and returns a __File object__.

File objects are a new type of data that represent a file on disk. They have useful methods, like strings (but unlike strings we can't really print them):

````python
my_file
````

### Reading file contents

`read()` is a File object method that returns the contents as a string. It has no arguments.

````python
my_file = open("weather.txt")
my_file.read()
````

````python
# usually we want to store the contents in a variable
my_file = open("weather.txt")
my_file_contents = my_file.read()
my_file_contents
````

````python
# or in one step...
my_file_contents = open("weather.txt").read()
my_file_contents
````

Remember the special character `\n`. Every line includes this new line character at the end. Remove it with the `rstrip()` method:

````python
my_file = open("weather.txt")
my_file_contents = my_file.read()

# remove the newline from the end of the file contents
my_record = my_file_contents.rstrip("\n")
my_record
````

````python
print('pythonnnn'.rstrip('nnn') )
print('pythonnnn'.rstrip('on') )
print('python'.lstrip('py') )
````

Notice how this version doesn't have the extra empty line.


Important: files are **exhaustible**. This means that once we have read the contents of a file, doing so again will give nothing:

````python
my_file = open("weather.txt")
my_file_contents1 = my_file.read()
my_file_contents2 = my_file.read()

(my_file_contents1, my_file_contents2)
````

This can easily cause confusion, so remember that you are only allowed to read the contents of a file once. This isn't actually a problem, since we can just store the contents of a file in a variable then use it multiple times. 

### Finding files

Sometimes Python can have a bit of trouble finding a file. In most environments, Python will look for the file in the same folder where the Python code file is saved. Some environments let you set this (Python calls it the *current working directory*) manually.


## Writing to files

To write to a file we have to use a second argument to open:

````python
# creating a new .txt doc
my_file = open("out.txt", "w")
````

`w` stands for write. Once we have opened a file for writing, we can use the `write()` method:

````python
my_file.write("Hello world")
````

How can we tell if this has worked? We need to open the file in a text editor (IDLE will do fine). 

## Closing files

Once we've finished writing data to a file, we have to close it:

````python
# if you can't close the file, it will be empty

my_file = open("out.txt", "w")
my_file.write("Hello world\n")
my_file.write("Hello world\n")
my_file.write("Hello CENTA")

# remember to close the file
my_file.close()
````

### Extracting part of a string

To get just part of a string, put square brackets after the variable name with the start and stop positions separated by a colon.

We start counting at __zero__ rather than one.
Positions are __inclusive__ at the start and __exclusive__ at the end.

Get positions 3 to 5 of the weather record:



````python
# my_record is already in memory
my_record
````

````python
# include the first integer position but not the last integer here - exclusive of the end
my_record[3:5]
````

There are some useful shortcuts. To go to the end of the string we just leave out the end position:

````python
my_record[5:]
````

The opposite to start from the begining:

````python
my_record[:8]
````

And a single number gives us a single character:

````python
my_record[3]
````

## Summary of all things!

|  __Name__ | __Job__  | __Argument__  | __Returns__  | __Type__  |
|---|---|---|---|---|
| `open()`  | opens a file for reading or writing  | filename, optional mode (both strings)  | File object  | built in function |
|  `read()` | reads the contents of a file  | none  | String  | method of File objects  |
| `rstrip()` | removes characters from end of string (usually newline)| string to remove  | string  | method of string objects |
| `write()`  | writes text to a file | string to write | nothing  | method of File objects |
|   `close()`| closes a file | none | nothing | method of File objects|
