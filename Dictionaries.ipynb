## Introducing paired data

Say we want to count all the rainy days in one of our weather strings:

````python
weather = "SRCSRCSRSRSRCSHRCSRCHSRCHSRCSRRSRHRCS"
rain_count = weather.count("R")
rain_count
````

That was pretty straightforward. How about all four weather types:

````python
weather = "SRCSRCSRSRSRCSHRCSRCHSRCHSRCSRRSRHRCS"
rain_count = weather.count('R')
snow_count = weather.count('S')
clear_count = weather.count('C')
hail_count = weather.count('H')

print(rain_count, snow_count, clear_count, hail_count)
````

Getting repetitive. How about two-day patterns (16 variables):

```python
rr_count = weather.count("RR")
rs_count = weather.count("RS")
rc_count = weather.count("RC")
...
```

or three-day patterns (64 variables):

```python
rrr_count = weather.count("RRR")
rrs_count = weather.count("RRS")
rrc_count = weather.count("RRC")
```

For any length of pattern, we can figure out how many unique patterns there are with a bit of maths:

````python
length_of_pattern = 10
4 ** 10
````

How might we store these weather pattern counts to avoid making lots of variables? Normally a list is the answer:

````python
weather = "SRCSRCSRSRSRCSHRCSRCHSRCHSRCSRRSRHRCS"

patterns = ['RR','RS','RH','RC',
            'SR','SS','SH','SC',
            'HR','HS','HH','HC',
            'CR','CS','CH','CC']

all_counts = []
for pattern in patterns:
    count = weather.count(pattern)
    all_counts.append(count)
    
print(all_counts)
print("the count for HS is " , all_counts[9])
````

````python
patterns.index('HS')
````

But you can see the problem: once they're stored in the list, there's no easy way to look up the count for a given pattern. There's no longer any connection between the patterns and the counts.

This is an example of paired data - also called key/value data.

Python's data structure for storing this type of data is a __dict__ (short for dictionary).

## Creating dicts

### Literal dicts

To make a dict 

- start and end with curly brackets
- separate keys and values with colons (label/IDs and  values)
- separate each pair (item) with a comma

````python
weather_codes = { 
'S' : 'Snow',
'R' : 'Rain',
'C' : 'Clear',
'H' : 'Hail'
}

weather_codes
````

We often write dicts on multiple lines. Getting a single value is similar to a list, but instead of giving the index, we give the key for the value we want:

````python
weather_codes['C']
````

### Building up a dict

We can create an empty dict, and add items to it one at a time:

````python
# create an empty dict
weather_codes = {}

# add one key/value pair at a time
weather_codes['C'] = 'Clear'
weather_codes['S'] = 'Snow'
weather_codes['R'] = 'Rain'

weather_codes
````

The thing that goes inside the square brackets is always the key, whether we are setting a value or retrieving a value. 

How does this help us with our weather patterns problem?

## Counting patterns with a dict

Here's how we store the counts in a dict. We start with an empty dict, and add one key/value pair for each dinucleotide:

````python
weather = "SRCSRCSRSRSRCSHRCSRCHSRCHSRCSRRSRHRCS"

patterns = ['RR','RS','RH','RC',
            'SR','SS','SH','SC',
            'HR','HS','HH','HC',
            'CR','CS','CH','CC']

all_counts = {}            # This time we create a dictionary rather than square brackets use curly
for pattern in patterns:
    count = weather.count(pattern)
    all_counts[pattern] = count      # Put pattern inside the syntax and then count to see labels/keys and values together
    
print(all_counts)
````

Notice how although it's bigger than our previous examples the `all_counts` dict has the same key/value structure. 

We can now look up the count (value) for a particular pattern (key) very easily:

````python
all_counts['RR']
````

### Removing zero counts

Problem: many of the counts are zero (and for longer patterns, nearly all the counts will be zero). Solution: just store the counts that are greater than zero:

````python
weather = "SRCSRCSRSRSRCSHRCSRCHSRRRSSSRCHSRCSRRSRHRCS"

patterns = ['RR','RS','RH','RC',
            'SR','SS','SH','SC',
            'HR','HS','HH','HC',
            'CR','CS','CH','CC']

all_counts = {}
for pattern in patterns:
    count = weather.count(pattern)
    if count > 0:
        all_counts[pattern] = count
    
all_counts
````

Now we are just storing the positive counts. This can lead to trouble when looking up counts for a pattern that doesn't occur in the sequence:

````python
all_counts['RR']
````

````python
all_counts['HH']
````

The `get()` method lets us specify a default for when the key isn't found:


````python
all_counts.get('HH', 0)
````

## Looping with dicts

The `keys()` method returns a list of all the keys in a dict:

````python
all_counts.keys()
````

There's also a `values()`:

````python
print(all_counts.values())
max(all_counts.values())
````

To get keys and values together, we can use `items()`:

````python
all_counts.items()
````

This is very useful for writing loops.
Note that we have to pick **two** variable names for the loop:

````python
# which patterns occur exactly twice in the sequence?

for pattern, count in all_counts.items():
    if count == 2:
        print(pattern)
````

````python
for two_day_pattern, two_day_count in all_counts.items():
    if two_day_pattern[0] == two_day_pattern[1]:
        print(two_day_pattern.lower(), two_day_count/sum(all_counts.values()))
````
