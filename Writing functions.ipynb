## Writing a simple function

Look at the wet-day content code from the start of the course:


````python
record = "RCRRSSCRCRSSCRR"

length = len(record)
rain_count = record.count('R')
snow_count = record.count('S')
wet_percentage = (rain_count + snow_count) / length * 100

print("Wet-day percentage is " + str(wet_percentage))
````

````python
def calculate_wet_day_percentage(record):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return wet_percentage

calculate_wet_day_percentage('RSRSRSRSRCCCRCRRSSCRCRSSCRR')
# calculate_wet_day_percentage(input(''))

records = ['RSRSCRRC' , 'SRSRSCCSRRRR' , 'SCSCSCRSRSCSRSCSRCSCRRR']
for r in records:
    print(calculate_wet_day_percentage(r))
````

````python
percentages = []
records = ['RSRSCRRC' , 'SRSRSCCSRRRR' , 'SCSCSCRSRSCSRSCSRCSCRRR']
for r in records:
    percentages.append(calculate_wet_day_percentage(r))

percentages
````

There's one line to define the input:

```python
record = "RCRRSSCRCRSSCRR"
```

and one line to define the output:

```python
print("Wet-day percentage is " + str(wet_percentage))
```

and only four lines to do the actual work:

```python
length = len(record)
rain_count = record.count('R')
snow_count = record.count('S')
wet_percentage = (rain_count + snow_count) / length * 100
```

We can turn this bit of code into a function that we call call:


````python
# define a new function
def get_wet_percentage(record):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return wet_percentage

# call the function
wet = get_wet_percentage("RCRRCSSRCRCRR")
print(wet)
````

Things to note about the definition line:

- the definition line starts with `def`
- next comes the name of our new function (arbirary)
- function argument names go in brackets (also arbitrary)
- def line ends with a colon
- body is indented
- use the argument names inside the body
- body ends with `return` to return the result

__Defining__ a function doesn't cause it to run; that only happens when we __call__ it. 

Argument variables (e.g. `record`) only exist inside the function.

When we write a function we don't have to worry about how it will be used - we just need to know the __inputs__ (arguments) and __output__.

When we use a function we don't have to worry about how it works inside - we just need to know the __inputs__ and __output__.

### Calling our new function

Once we've written our function, we can use it in lots of different ways:


Calulate the wet-day content of a record in a file
```python
record = open('weather.txt').read()
wet = get_wet_percentage(record)
```

Print the wet-day content for a given record
```python
print(get_wet_percentage("RCRRCR"))
```

Write the wet-day content for a record to a file
```python
result = open('output.txt', 'w')
result.write(str(get_wet_percentage('RCCRRCR')))
```

This separation of code is very valuable - __encapsulation__. 



## Things to avoid when writing functions
### No input

We can write a function that relies on variables defined outside it rather than arguments:

````python
record = "RCRRCR"

def get_wet_percentage():
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return wet_percentage


wet = get_wet_percentage()
print(wet)
````

This breaks encapsulation - now we have to know what variables are set in order to write the function. 

### No output

We can write a function that prints the value instead of returning it:

````python
def get_wet_percentage(record):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    print(wet_percentage)

get_wet_percentage("RCRRCRSSCRCR")
````

but this also breaks encapsulation - we have to know how the function will be used in order to write it. 

A good rule of thumb: _information gets in by arguments, information gets out by return value_.

## Improving our function
### Adding another argument

One problem currently is that we get too many decimal places:

````python
def get_wet_percentage(record):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return wet_percentage

get_wet_percentage("RCRRCRCRCRRR")
````

We can get round this by calling the `round()` function on the answer before we return it:

````python
round(1.23456789, 3)
````

````python
def get_wet_percentage(record):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return round(wet_percentage, 2)
                 
get_wet_percentage("RCRRCRCRCRRR")
````

What if we want more/fewer decimal places? Make the argument to `round()` an argument of our function:

````python
def get_wet_percentage(record, decimal_places):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return round(wet_percentage, decimal_places)
                 
get_wet_percentage("RCRRCRCRCRRR", 7)
````

````python
get_wet_percentage("RCRRCRCRCRRR", 3)
````

### Default values for function arguments

In many cases, we don't really care about picking a number of decimal places. We can add a default to the definition:

````python
def get_wet_percentage(record, decimal_places=2):
    length = len(record)
    rain_count = record.count('R')
    snow_count = record.count('S')
    wet_percentage = (rain_count + snow_count) / length * 100
    return round(wet_percentage, decimal_places)
                 
get_wet_percentage("RCRRCRCRCRRR")
````

````python
get_wet_percentage("RCRRCRCRCRRR", 4)
````

````python
# for testing try easier, shorter inputs
get_wet_percentage("RC")
````

````python
get_wet_percentage("RCRRCRCRCRRR", 4)
````

````python
get_wet_percentage("RCRRCRCRCRRR", 4)
````

### Keyword arguments

In all the examples above, we supply the arguments in the same order as the definition. If we want to use a different order (or just be more explicit) we can use keyword arguments:

````python
get_wet_percentage(decimal_places=3, record="RCRRCRCRCRRR")
````

## Testing functions

When we're working on a new function, we might want to test if it's working correctly. Use `assert` with a condition:

````python
assert get_wet_percentage("R") == 100
assert get_wet_percentage("C") == 0
assert get_wet_percentage("RSCC") == 50
assert get_wet_percentage("RCC", 1) == 33.3
assert get_wet_percentage("RCC", 5) == 33.33333

# if something wrongs assert will show an error
````

If an `assert` is false, then it will stop and print an error message.

Assertions are good for:

- providing some documentation about the behaviour of the function
- reassuring you that your function is giving the right answer
- letting you know if you break the function 
- demonstrating how the function can be used to other people

and also for

- making it easy to me to write exercises!

### Condition percentage, part one

Write a function that takes two arguments – a weather record string and a condition code – and returns the percentage of days that had that weathe. Use the following assertions to test your function:

```python
assert my_function("RRCSRHCCSSCCRCCCSRRC", "H") == 5
assert my_function("RRCSRHCCSSCCRCCCSRRC", "R") == 30
assert my_function("RRCSRHCCSSCCRCCCSRRC", "c") == 45
assert my_function("RRCSRHCCSSCCRCCCSRRC", "W") == 0
```

Remember to change the name of the function in the `assert` statements to match your function name. 
    


````python
def calculate_weather_percentage_days(record, weather):
    
    # make the input case sensitive
    weather = weather.upper()
    record = record.upper()
    
    weather_count = record.count(weather)
    weather_percentage = weather_count / len(record) * 100
    
    return weather_percentage

assert calculate_weather_percentage_days("RRCSRHCCSSCCRCCCSRRC", "H") == 5
assert calculate_weather_percentage_days("RRCSRHCCSSCCRCCCSRRC", "R") == 30
assert calculate_weather_percentage_days("RRCSRHCCSSCCRCCCSRRC", "c") == 45
assert calculate_weather_percentage_days("RRCSRHCCSSCCRCCCSRRC", "W") == 0
````

### Condition percentage, part two

Modify the function from part one so that it accepts a list of condition codes rather than a single one. If no list is given, the function should return the percentage of days that were either rainy or windy (`R` and `S`). Your function should pass the following assertions:

```python
assert my_function("RRCSRHCCSSCCRCCCSRRC", ["C"]) == 40
assert my_function("RRCSRHCCSSCCWWCCSRRC", ["R","S","H"]) == 50
assert my_function("RRCSRHCCSSCCWWCCSRRC") == 35
```

To get this one to work, you'll have to go through the list of condition codes one at a time, generate the count for each one, and come up with a total count.


````python
record = "RRCSRHCCSSCCWWCCSRRC"
codes = ["R","S","H"]

total = 0
for code in codes:
    total = total + record.count(code)
    print(code, record.count(code), total)

total_percentage = total / len(record) * 100
total_percentage
````

````python
def multiple_condition_percentage(record, codes=['R','S']):

    total = 0
    for code in codes:
        total = total + record.count(code)
        # print(code, record.count(code), total)
    
    total_percentage = total / len(record) * 100
    
    return total_percentage

assert multiple_condition_percentage("RRCSRHCCSSCCRCCCSRRC", ["C"]) == 45
assert multiple_condition_percentage("RRCSRHCCSSCCWWCCSRRC", ["R","S","H"]) == 50
assert multiple_condition_percentage("RRCSRHCCSSCCWWCCSRRC") == 45
````
