## Lists
### Defining a list and getting elements
Python's solution is to have lists of data, which can store multiple values. Lists start and end with square brackets and the __elements__ are separated by commas:

````python
headers = ["ABC123","DEF456","HIJ789"]
headers
````

Once we have stored some values in a list, we can get a single element by giving its __index__:

````python
headers[0]
````

Indexes work just like they did for strings:

````python
headers[1:3]
````

Start counting from zero, inclusive at the start, exclusive at the end. 

### Building up a list

Often we want to start with an empty list and build it up:

````python
headers = []
````

and build it up. `append()` adds a value on to the end of the list:

````python
headers = []
print(headers)
headers.append("ABC123")
headers.append("climate")
headers.append("10.50")
print(headers)
````

all of the list methods (including .append) change the original

````python
s = 'banana'
x = s.upper()
print(s)
print(x)
````

````python
headers.reverse()
headers
````

````python
headers + ["DEF456","HIJ789"]
````

note that this doesn't change the original list:

````python
headers
````

### Creating a list by splitting a string

The `split()` method of strings returns a list:

````python
sentence = "one two three four"
sentence.split(" ")
````

This doesn't change the string:

````python
sentence
````

````python
renewal = "urban regeneration policies"
renewal.split('n')
````

Another way to create a list is by using the `range()` function. `range()` can generate lists of numbers. In Python 3, `range()` has a special behaviour that we call __lazy__ - it only generates the numbers as we use them. This means that if we want to see the numbers printed, we have to call `list()` on the result of `range()` (as below). When we want to use a range in a real program, we don't have to to this though.

With one argument it counts from zero to just before the number:

````python
list(range(10))
````

with two arguments it counts from the first number to just before the second:

````python
list(range(5,12))
````

With three arguments, it uses the third argument as the step size:

````python
list(range(11, 20))
````

````python
list(range(3, 24, 4))
````

````python
list(range(1000, 2000, 200))
````

## Loops

We know how to do something for one element of a list:

````python
headers = ["DEF456","ABC123","HIJ789"]
one_header = headers[1]
print(one_header)
len(one_header)
````

How do we do something for every element of a list?

````python
for header in headers:
    print("one header is " + header + " " + "hello" + " " + header)
    print("hello")
````

````python
for country_ISO in headers:
    print("the ISO is " + country_ISO)
````

 - `header` is a new variable that only exists inside the loop
 - The `for` line ends with a colon
 - The body of the loop is indented (tabs or spaces but not both!)
 - The value of header is set to each element in turn:
 
 We can do all the stuff that we already know how to do inside a loop.

````python
transects = ['fgwufgwfgw', 'gwwfug', 'wufgfgwugf']
transects.append('fgfgfgfgfuwwffguugug')
for transect in transects:
    print("one transect is " + transect)
    print("it starts with " + transect[0])
    print("its length is " + str(len(transect)))
    print("it contains " + str(transect.count('w')) + ' water cells ')
    print()
````

````python
transects
````

### Loops and strings

We can write a loop that uses a string as if it were a list - each character will behave like an individual element:

````python
wind = "NNEESSWWNNW"

# strings can pretend to be lists
# each element is a character
for direction in wind:
    print(direction)
````

It's easy to do this by accident so if you see individual characters, check you're looping over the right value. 

### Loops and files

We can treat a file object as if it were a list, each line is a list element:

````python
my_file = open("test.txt")

# files can pretend to be lists
# each element is a line
for fruit in my_file:
    # print the second letter of each line
    print(fruit[1:4])
    print(fruit.count('a'))
````

````python
# do not use .read() here as it reads the document to as strings
my_file = open("test.txt").read()

for x in my_file:
   
    print(x[0])
    print(x.count('a'))
````

This is great for all sorts of file processing. For example, how many letters are in each line of my file?

````python
for line in open('test.txt'):
    fruit = line.rstrip('\n')
    print(len(fruit))
````

Remember that files are **exhaustible**, so we can't both read the contents of a file and loop over it:

````python
my_file = open("test.txt")
my_file_contents = my_file.read()
# now my_file is exhausted

# this loop never runs because we are already at the end of the file
for line in my_file:
    print(line[1])
````

and we can't loop over a file twice:

````python
my_file = open("test.txt")

for line in my_file:
    print(line[1])

# this loop never runs because we are already at the end of the file
for line in my_file:
    fruit = line.rstrip('\n')
    print(len(fruit))
````

If we need to do multiple things to each line in a file, we just put them inside one big loop:

````python
# with better variable names

my_file = open("test.txt")

for line in my_file:
    second_character = line[1]
    number_of_characters = len(line)
    print(second_character, number_of_characters)
````
