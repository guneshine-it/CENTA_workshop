## True and False

__Conditions__ are things that can be evaluated as either true or false:

````python
# operators
print(3 == 5)
print(3 > 5)
print(3 <= 5)
print(5 != 4)

print(len("NWWSSE") > 5)
print("NWWSSE".count("S") > 1)

# methods
print("WIND".startswith("WI"))
print("WIND".endswith("ND"))
print("WIND".isupper())
print("WIND".islower())

print("N" in ["N", "E", "S", "W"])
````

There are a bunch of different types of true/false conditions in Python:

- equals (and not equals with `!=`)
- numerical comparisons (`>`, `<`, `>=`, `<=`)
- string methods (`startswith`, `isupper`)
- is a value in a list

We can experiment by typing in at the console:

````python
4 != 5
````

## Using conditions

### `if` statements

The simplest thing we can do with a condition is execute some code if it's true:

````python
rainfall_mm = 90
if rainfall_mm > 100:
    print("heavy rainfall day")

print('hello')
# if you write 90, it doesn't show anything
# as print('hello') not in the loop, the output shows it
````

Notice that

- the condition line starts with `if`
- the thing we want to test goes after the `if`
- the line ends with a colon
- just like with loops, the body is indented
- can have multiple lines in the body

A more interesting example:

````python
records = ['rrcscr', 'rsrsrcrssrcrsrsrr', 'ccrsrcc', 'rrsrcrcrcsrsrcr']
for record in records:
    if record.count("r") > 2:
        print(record + " contains more than two rain tokens")
````

````python
records = ['rrcscr', 'rsrsrcrssrcrsrsrr', 'ccrsrcc', 'rrsrcrcrcsrsrcr']
for record in records:
    if record.count("r") > record.count("c"):
        print(record.upper() + " contains more rainy days than clear days")
````

Code with conditions often involves mulitple levels of indentation. Watch out for `IndentationErrors` and indentation mistakes.

### `else` statements

The `if` examples above are yes/no - either we execute the bit of code, or we do nothing. 

Sometimes we want either/or - two different branches:

````python
records = ['rrcscr', 'rsrsrcrssrcrsrsrr', 'ccrsrcc', 'rrsrcrcrcsrsrcr']
for record in records:
    if record.startswith('r'):
        print(record  + " starts with rain")
    else:
        print(record  + " doesn't start with rain")
````

The `else` line has nothing on it. The `else` block is indented and is run when the condition is false.

---

### `elif` statements

Sometimes we need more than two branches. Use `elif` for each extra branch:

````python
records = ['rrcscr', 'rsrsrcrssrcrsrsrr', 'ccrsrcc', 'srsrcrcrcsrsrcr']
for record in records:
    if record.startswith('r'):
        print(record  + " starts with r")
    elif record.startswith('c'):
        print(record  + " starts with c")
    else:
        print(record  + " starts with something else")
````

We can add as many conditions as we like:

````python
records = ['rrrsrc', 'ccccsc', 'crscrr', 'cccrrs', 'rrsrcc', 'scrcsc', 'hrcssc']
for record in records:
    if record.startswith('r'):
        print('r!')
    elif record.startswith('c'):
        print('c!')
    elif record.startswith('s'):
        print('s!')
    else:
        print('ERROR: ' + record)
````

`else` and `elif` are good when we have _mutually exclusive_ posibilities. If more than one can be true, then use multiple `if` lines:

````python
records = ['rrrsrc', 'ccccss', 'crscrr', 'cccrrs', 'rrsrch', 'scrcsc', 'hrcssc']

for record in records:
    if record.startswith('r'):
        print(record  + " starts with r")
    elif record.endswith('c'):
        print(record  + " ends with c")
````

Compare the output with what we get if we use `elif`

## Complex conditions
### `and`

We can join together two conditions with `and` to make a complex one that's only true if both of the simple ones are true:

````python
records = ['rrrsrc', 'ccccss', 'crscrr', 'cccrrs', 'rrsrch', 'scrcsc', 'hrcssc']
for record in records:
    if record.startswith('r') and record.count('r') > 3:
        print("both conditions are true for " + record)
````

### `or`

We can do the same with `or` and it will be true if either of the simple conditions are true:

````python
records = ['rrrsrc', 'ccccss', 'crsrrr', 'cccrrs', 'rrsrch', 'scrcsc', 'hrcssc']
for record in records:
    if record.startswith('r') or record.count('r') > 3:
        print("one condition is true for " + record)
````

We can even join together complex conditions in this way:

````python
records = ['rrrsrc', 'ccccss', 'crsrrr', 'cccrrs', 'rsssch', 'scrcsc', 'hrcssc']
for record in records:
    if (record.startswith('r') or record.startswith('c')) and record.count('r') > 2:
        print("the condition is true for " + record)
````

Note that we need the parens here to make it

    (A or B) and C
    
without them, Python will interpret it as

    A or (B and C)
