### Comments
To include a bit of text for humans to read, start with `#`. We call this a __comment__:

```python
# print a friendly greeting
print("Hello martin")
```

Some characters in Python have a special meaning: for example `\n` means _start a new line_.

```python
# print the same greeting, over two lines
# print your name and surname in the same way
print("Hello\nworld") 
print("Bengi\nGunes")
```

### Variables
To store a bit of text to use later, we use a __variable__:

```python
# store a weather station code in the variable my_station
my_station = "ABQ01"

# now print the station code
print(my_station)

# change the value of my_station
my_station = "SEA02"

# print it again
print(my_station)
```

Variable names can be whatever we want - this code is exactly the same as the previous one:

```python
# store a station code in the variable banana
banana = "ABQ01"

# now print the station code
print(banana)
```

```python
# change the value of banana
banana = "SEA02"

# print it again
print(banana)
```

## Manipulating text (strings)
### Concatenation
We can join two strings together with `+`:

```python
'Hello ' + 'World'
```

__Important: anywhere we can use a string written with quotes, we can also use a variable that stores a string:__

```python
region = "COAST"
label = region + "_TEMP"
label
```

```python
region = "COAST"
season = "SUMMER"
label = region + "_TEMP_" + season
label
```

The result of concatenating two (or more) strings is also a string:

```python
print("Hello" + " " + "world")
print("Elif" + " " + "Yerlikaya")
```

### Lengths

To find the length of a string, use the `len()` function:

```python
len("yerlikaya")
len("bengi")
len("elif")
```

What you need to know about `len()`:

- `len()` is a function
- it takes one argument
- the argument has to be a string
- it returns the length of the string
- we have to do something with the returned value
- __the length is a number, not a string!__

Numbers and strings are different:

```python
# this works and gives addition
2 + 2
```

```python
# this works and gives concatenation
"abc" + "def"
```

```python
# this causes an error
2 + "abc"
```

### Changing case

To turn a string into lower case, use a __method__ called `lower()`:

```python
site_code = "MNTN"

# make a version in lower case
site_code.lower()
```

```python
new_code = "BTWIN"

# another version
new_code.lower()
```

Look closely at how this works:

- `lower()` is a __method__, not a __function__
- we write the name of the variable first, then a dot, then the method name
- `lower()` has no arguments
- `lower()` only works on strings, not numbers
- __`lower()` doesn't change the variable, it returns a lower case version__

```python
site_code = "RIDGE"

# print the variable
print("before: " + site_code)

# run the lower method and store the result
lowercase_code = site_code.lower()
print(lowercase_code)

# print the variable again
print("after: " + site_code)

# print the variable again
print("after: " + lowercase_code)
```

## Find and replace in strings

Use the `replace()` method:

```python
forecast = "sunny and warm"
print(forecast) 

# replace sunny with foggy
print(forecast.replace("sunny", "foggy"))

# we can replace more than one character
print(forecast.replace("warm", "cool"))

# the original variable is not affected
print(forecast)
```

```python
# method chaining
print(forecast.replace("sunny", "foggy").replace("warm", "cool"))
```

```python
forecast = "sunny and warm"

new_forecast = forecast.replace("sunny", "foggy")

forecast, new_forecast
```

```python
my_log = "rrddsrddrddsddrrsrddr"

# count weather events
rain_count = my_log.count('r')
dry_count  = my_log.count('d')
snow_count = my_log.count('s')
hail_count = my_log.count('h')

 
# now print the counts
print("rain: " + str(rain_count))
print("dry: "  + str(dry_count))
print("snow: " + str(snow_count))
print("hail: " + str(hail_count))
```

### Finding positions of substrings

The `find()` will tell you the position of a short string inside a long one:

```python
my_log = "rrddsrddrddsddrrsrddr"

print(my_log.find('r'))
print(my_log.find('d'))
print(my_log.find('h'))
```

Remember:

- we start counting from zero
- if the short string isn't there, we get `-1`
